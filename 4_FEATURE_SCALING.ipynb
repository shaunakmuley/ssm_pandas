```python
# These ar imp libraries for graph and scaling and normalization
from scipy import stats
from mlxtend.preprocessing import minmax_scaling
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# FEATURE SCALING
# Transforming data so that it fits into specific range 
# Bringing all columns in one scale
# Think scaling like changing size of the photo so it can fit in the collage

# normalization features between zero and 1 
# standardization feature between centered around zero


# Normalization min_max_scaling
# Standardization 
```

```python
# "normalize" has two completely different meanings depending on who you are talking to:

# Machine Learning meaning : "Normalize" means squishing the data into 
# a strict 0-to-1 range using Min-Max scaling.

# Statistics meaning : "Normalize" means physically transforming the data
# so its distribution becomes a Normal distribution (a bell curve).
```

```python
#  NORMALIZATION
# Machine Learning meaning

# generate 1000 data points randomly drawn from an exponential distribution
original_data = np.random.exponential(size=1000)

# mix-max scale the data between 0 and 1
scaled_data = minmax_scaling(original_data, columns=[0])

# plot both together to compare
fig, ax = plt.subplots(1, 2, figsize=(15, 3))
sns.histplot(original_data, ax=ax[0], kde=True, legend=False)
ax[0].set_title("Original Data")
sns.histplot(scaled_data, ax=ax[1], kde=True, legend=False)
ax[1].set_title("Scaled data")
plt.show()
```

```python
# original_data = np.random.exponential(size=1000)
# This code generates 1000 different data points 

# scaled_data = minmax_scaling(original_data, columns=[0])
# By this code we can convert original data into scaled data values between 0 and 1 

```

```python
# NORMALIZATION
# Statistics meaning
# It is the type of the feature scaling 

# normalize the exponential data with boxcox
normalized_data = stats.boxcox(original_data)

# plot both together to compare
fig, ax=plt.subplots(1, 2, figsize=(15, 3))
sns.histplot(original_data, ax=ax[0], kde=True, legend=False)
ax[0].set_title("Original Data")
sns.histplot(normalized_data[0], ax=ax[1], kde=True, legend=False)
ax[1].set_title("Normalized data")
plt.show()
```
