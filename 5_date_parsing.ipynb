```python
# modules we'll use
import pandas as pd
import numpy as np
import seaborn as sns
import datetime
```

```python
# In this we will see how to convert date to proper date format
```

```python
file= pd.read_csv("catalog.csv")
# set seed for reproducibility
np.random.seed(0)


# Without np.random.seed(0): 
# Every time you run your script, you get 1,000 brand new numbers. Your histograms will shift and change shape slightly every time you run it.

# With np.random.seed(0): 
# You will get the exact same 1,000 numbers every single time you hit run. Your graphs will never change.
```

```python
# create a new column, date_parsed, with the parsed dates
file['date_parsed'] = pd.to_datetime(file['date'], format="%m/%d/%y")
```

```python
file['date_parsed'].head()
# In this dtype is datetime
```

```python
file['date_parsed'].dt.day.head()
# In this date is printed of the respective row in dataset
```

```python
file['date_parsed'].dt.month.head()
```

```python
file['date_parsed'].dt.year.head()
```

```python
dr=file['date_parsed'].dropna()
# you dropped null values of date
```

```python

# plot the day of the month
sns.distplot(dr, kde=False, bins=31)
```
